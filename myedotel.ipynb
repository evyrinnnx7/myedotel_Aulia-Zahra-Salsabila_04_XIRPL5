```python
# Bagian 2. Abstract Class KamarHotel
from abc import ABC, abstractmethod


class KamarHotel(ABC):

    def __init__(self, nama_kamar, stok, harga_dasar):
        self.__nama_kamar = nama_kamar
        self.__stok = stok
        self.__harga_dasar = harga_dasar

    def get_nama_kamar(self):
        return self.__nama_kamar

    def get_stok(self):
        return self.__stok

    def get_harga_dasar(self):
        return self.__harga_dasar

    def tambah_stok(self, jumlah):
        if jumlah < 0:
            print(
                f"Gagal update stok {self.__nama_kamar}! "
                f"Stok tidak boleh negatif ({jumlah})."
            )
        else:
            self.__stok += jumlah
            print(
                f"Berhasil menambahkan stok {self.__nama_kamar}: "
                f"{jumlah} unit."
            )

    @abstractmethod
    def tampilkan_detail(self):
        pass

    @abstractmethod
    def hitung_harga_total(self, jumlah_malam):
        pass
```

```python
# Bagian 3. Kamar Deluxe
class KamarDeluxe(KamarHotel):

    def __init__(self, nama_kamar, stok, harga_dasar, fasilitas):
        super().__init__(nama_kamar, stok, harga_dasar)
        self.fasilitas = fasilitas

    def tampilkan_detail(self):
        print(
            f"[DELUXE] {self.get_nama_kamar()} | "
            f"Fasilitas: {self.fasilitas}"
        )

    def hitung_harga_total(self, jumlah_malam):
        harga_dasar = self.get_harga_dasar()
        pajak = harga_dasar * 0.10

        return (harga_dasar + pajak) * jumlah_malam
```

```python
# Bagian 4. Kamar Standard
class KamarStandard(KamarHotel):

    def __init__(self, nama_kamar, stok, harga_dasar, kapasitas):
        super().__init__(nama_kamar, stok, harga_dasar)
        self.kapasitas = kapasitas

    def tampilkan_detail(self):
        print(
            f"[STANDARD] {self.get_nama_kamar()} | "
            f"Kapasitas: {self.kapasitas}"
        )

    def hitung_harga_total(self, jumlah_malam):
        harga_dasar = self.get_harga_dasar()
        pajak = harga_dasar * 0.05

        return (harga_dasar + pajak) * jumlah_malam
```

```python
# Bagian 5. Fungsi Transaksi
def proses_transaksi(daftar_pesanan):

    print("\n--- STRUK PEMESANAN ---")

    total_tagihan = 0

    for pesanan in daftar_pesanan:

        kamar = pesanan["kamar"]
        jumlah_malam = pesanan["malam"]

        kamar.tampilkan_detail()

        harga_dasar = kamar.get_harga_dasar()

        if isinstance(kamar, KamarDeluxe):
            pajak_persen = 10
        else:
            pajak_persen = 5

        pajak = harga_dasar * pajak_persen / 100
        subtotal = kamar.hitung_harga_total(jumlah_malam)

        print(
            f" Harga Dasar/Malam: Rp {harga_dasar:,.0f} | "
            f"Pajak({pajak_persen}%): Rp {pajak:,.0f}"
        )

        print(
            f" Menginap: {jumlah_malam} malam | "
            f"Subtotal: Rp {subtotal:,.0f}"
        )

        total_tagihan += subtotal

        print()

    print("----------------------------------------")
    print(f"TOTAL TAGIHAN: Rp {total_tagihan:,.0f}")
    print("----------------------------------------")
```

```python
# Bagian 6. Setup Data Kamar
print("--- SETUP DATA KAMAR ---")

kamar_deluxe = KamarDeluxe(
    "Kamar Deluxe Sea View",
    0,
    1500000,
    "Private Pool"
)

kamar_standard = KamarStandard(
    "Kamar Standard Superior",
    0,
    500000,
    "2 Orang"
)
```

```python
# Bagian 7. Tambahkan Stok
kamar_deluxe.tambah_stok(10)

kamar_standard.tambah_stok(-5)

kamar_standard.tambah_stok(20)
```

```python
# Bagian 8. Buat Pesanan
daftar_pesanan = [
    {
        "kamar": kamar_deluxe,
        "malam": 2
    },
    {
        "kamar": kamar_standard,
        "malam": 1
    }
]
```

```python
# Bagian 9. Jalankan Transaksi
proses_transaksi(daftar_pesanan)
```


**Analisis Proyek**
